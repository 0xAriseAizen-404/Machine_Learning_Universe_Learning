## DBSCAN Clutering

```python
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.datasets import make_circles
import numpy as np
```

```python
X, y = make_circles(n_samples=750, factor=0.3, noise=0.1)
```

```python
plt.scatter(X[:,0],X[:,1])
```

```python
from sklearn.cluster import DBSCAN
```

```python
dbcan=DBSCAN(eps=0.1)
```

```python
dbcan.fit_predict(X)
```

```python
dbcan.labels_
```

```python
labels_fitpredict = dbcan.fit_predict(X)
labels_attribute = dbcan.labels_
np.array_equal(labels_fitpredict, labels_attribute)
```

```python
plt.scatter(X[:,0],X[:,1],c=dbcan.labels_)
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```
