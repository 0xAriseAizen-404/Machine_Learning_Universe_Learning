```python
# Importing Saved Model
from gensim.models import Word2Vec

# Load the saved model
model = Word2Vec.load("./word2vec_amazon_cell_reviews.model")
```

```python
from sklearn.decomposition import PCA
import random
import plotly.express as px

idxs = random.sample(range(len(model.wv.index_to_key)), 100)

X = PCA(n_components=3).fit_transform(
    model.wv.get_normed_vectors()[idxs]
)

words = [model.wv.index_to_key[i] for i in idxs]

fig = px.scatter_3d(
    x=X[:, 0],
    y=X[:, 1],
    z=X[:, 2],
    color=words,
    # hover_name=words,
)

fig.show()
```
