```python
# Importing Saved Model
from gensim.models import Word2Vec

# Load the saved model
model = Word2Vec.load("./word2vec_amazon_cell_reviews.model")
```

```python
model.wv.most_similar("bad")
```

```python
model.wv.similarity('cheap', 'inexpensive')
```

```python
model.wv.similarity('good', 'nice')
```

```python
print(model.wv["battery"].shape)
model.wv.most_similar("battery")
```

```python
model.wv.most_similar(model.wv["king"] - model.wv["man"] + model.wv["woman"])
```

```python
word = "queen"

if word in model.wv:
    print("Word exists")
else:
    print("Word not found")
```

```python
# Vocabulary Size
print(len(model.wv.index_to_key))
```

```python
print(model.wv.get_vecattr("king", "count"))
print(model.wv.get_vecattr("queen", "count"))
print(model.wv.get_vecattr("man", "count"))
print(model.wv.get_vecattr("woman", "count"))
```

```python
# Word Analogy
# A : B :: C : D
model.wv.most_similar(
    positive=["earphones", "cable"],
    negative=["charger"]
)
```
